```python
import pulp

items = ["p1", "p2", "p3", "p4"]

values = {
    "p1": 1,
    "p2": 4,
    "p3": 3,
    "p4": 3
}

weights = {
    "p1": 13,
    "p2": 12,
    "p3": 8,
    "p4": 10
}

limit = 30

# Create a LP Maximization problem
prob_knapsack = pulp.LpProblem("Knapsack", pulp.LpMaximize)

# Variables
x = pulp.LpVariable.dict('x', items, lowBound=0, upBound=1, cat=pulp.LpInteger)

# Objective (set once)
prob_knapsack += sum(values[i] * x[i] for i in items)

# Constraint
sum_weights = 0
for i in items:
    sum_weights += weights[i] * x[i]

prob_knapsack += sum_weights <= limit

# Different (cleaner) way to write the constraint
# prob_knapsack += sum(weights[i] * x[i] for i in items) <= limit

# Optimize the model
prob_knapsack.solve()

# Print the status of the solved LP
print("Status = %s" % pulp.LpStatus[prob_knapsack.status])

# Print the value of the variables at the optimum
for i in items:
    print("%s = %f" % (x[i].name, x[i].varValue))

# Print objective and constraints
print(prob_knapsack.objective)

# Print constraints values
for name in prob_knapsack.constraints.keys():
    print(f"{name} : {prob_knapsack.constraints[name].value()}")
```
